# **1. Définition et Modalités en Imagerie Médicale**
Une image médicale est la matérialisation sous forme d’images d’informations anatomiques ou fonctionnelles *in vivo* de parties (organes, tissus, cellules) du corps humain [BIHAN 2018][cite: 5].

Dans cet atelier, nous allons manipuler trois modalités différentes :
1. **Radiographie thoracique (Rayons X)** : Idéale pour observer les structures denses (os, opacités pulmonaires).
2. **IRM Cérébrale** : Idéale pour l'étude fine des tissus mous du cerveau.
3. **Endoscopie** : Observation directe des muqueuses et des mouvements organiques (ex: cordes vocales)[cite: 5].
